```python
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader

from traffic_signs.dataset import BelgiumTSDataset
from traffic_signs.transforms import get_transforms

train_ds = BelgiumTSDataset("../data/Training", transform=get_transforms(train=True))
print("Aantal trainingsafbeeldingen:", len(train_ds))

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True, num_workers=0)
images, labels = next(iter(train_loader))
print("Batch afbeeldingen:", images.shape)
print("Batch labels:", labels.shape, labels[:10])
```
