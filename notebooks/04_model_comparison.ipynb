```python
import json
import matplotlib.pyplot as plt

with open("../runs/baseline_cnn/history.json") as f:
    cnn_history = json.load(f)
with open("../runs/resnet18/history.json") as f:
    resnet_history = json.load(f)

plt.figure(figsize=(8, 4))
plt.plot(range(1, len(cnn_history["val_acc"]) + 1), cnn_history["val_acc"], label="Baseline CNN")
plt.plot(range(1, len(resnet_history["val_acc"]) + 1), resnet_history["val_acc"], label="ResNet18 (frozen)")
plt.xlabel("epoch")
plt.ylabel("validation accuracy")
plt.title("Baseline CNN vs. ResNet18 transfer learning")
plt.legend()
plt.show()

print(f"Beste CNN val-accuracy:    {max(cnn_history['val_acc']):.3f}")
print(f"Beste ResNet val-accuracy: {max(resnet_history['val_acc']):.3f}")
```
